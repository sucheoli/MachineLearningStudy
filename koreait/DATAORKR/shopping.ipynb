```python
import pandas as pd

df = pd.read_csv("data/소상공인시장진흥공단_상가(상권)정보_서울_202510.csv", low_memory=False)

df
```

```python
pd.set_option('display.max_column', 40)

df
```

```python
df.info()
```

```python
df.columns
view_columns =  ['상호명', '지점명', '상권업종대분류명', '상권업종중분류명',
                '상권업종소분류명', '시도명', '시군구명', '도로명', '도로명주소',
                '경도', '위도']

df = df[view_columns]
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] ='Malgun Gothic'
plt.rcParams['axes.unicode_minus'] =False
```

```python
import matplotlib.pyplot as plt

# 산점도 그리기
plt.figure(figsize=(6, 8))  # 그래프 크기 설정
plt.scatter(df['경도'], df['위도'])  # 산점도 그리기
plt.xlabel('경도')  # x축 레이블
plt.ylabel('위도')  # y축 레이블
plt.grid(True)  # 그리드 추가
plt.show()  # 그래프 출력|
```

```python
import seaborn as sns
plt.figure(figsize = (10, 12))
sns.scatterplot(data = df, x = '경도', y = '위도', hue = "시군구명")
```

```python
shop_edu = df[df['상권업종대분류명'] == '교육']
shop_edu.shape
```

```python
plt.figure(figsize = (10,12))

sns.scatterplot(data=shop_edu, x='경도', y = '위도', hue = '상권업종중분류명')
```

```python
shop_edu_etc = shop_edu[shop_edu['상권업종중분류명'] == '기타 교육']

shop_edu_etc.shape
```

```python
plt.figure(figsize = (10,12))

sns.scatterplot(data=shop_edu_etc, x='경도', y = '위도', hue = '상권업종소분류명')
```

```python
shop_edu_etc_com = shop_edu_etc[shop_edu_etc['상권업종소분류명'] == '컴퓨터 학원']

shop_edu_etc_com.shape
```

```python


plt.figure(figsize = (12,10))

sns.scatterplot(data = shop_edu_etc_com, x = '경도', y = '위도', hue = '시군구명')
```
